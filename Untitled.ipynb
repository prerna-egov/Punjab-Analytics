```python
import os
import pandas as pd
from datetime import datetime
import json
import gc


property_df = pd.read_csv(
    '/home/prerna/Punjab/punjab-data-prod-analysis/phagwara/eg_pt_property_phagwara_v2.csv',
    usecols=['id', 'propertyid', 'additionaldetails']
)

defaulter_report = pd.read_csv('/home/prerna/Downloads/Punjab_Data_Analysis_Phagwara_PDC_FINAL_amountDue_1.csv')

print(property_df.shape)
print(defaulter_report.shape)
```

```python
def get_year_construction(val):
    if pd.isna(val): return None
    try: return json.loads(val).get('yearConstruction')
    except: return None

property_df['YearOfConstruction'] = property_df['additionaldetails'].apply(get_year_construction)

print(property_df)
```

```python
# Rename column for easier merging
property_df.rename(columns={'propertyid': 'Property ID'}, inplace=True)

# Perform a left merge: keep all rows in defaulter_report
merged_df = defaulter_report.merge(
    property_df[['Property ID', 'YearOfConstruction']],
    on='Property ID',
    how='left'
)

# Optional: check the result
print(merged_df.head())
print(merged_df.shape)
```

```python
merged_df.to_csv('Defaulter_report_with_yoc.csv', index=False)
```
